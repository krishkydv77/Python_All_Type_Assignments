```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
df=pd.read_csv('/content/tips.csv')
df.head(5)
```

```python
df.isnull().sum()
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.drop('tip',axis=1),df['tip'],test_size=0.3,random_state=2)
```

```python
si=SimpleImputer(strategy='mean')
x_train_totalbill=si.fit_transform(x_train[['total_bill']])
x_test_totalbill=si.transform(x_test[['total_bill']])
x_test_totalbill.shape
```

```python
df['day'].value_counts()
```

```python
oe=OrdinalEncoder(categories=[['Thur','Fri','Sat','Sun']])
x_train_day=oe.fit_transform(x_train[['day']])
x_test_day=oe.transform(x_test[['day']])
x_test_day.shape
```

```python
ohe=OneHotEncoder(sparse_output=False,drop='first')
x_train_smoker=ohe.fit_transform(x_train[['smoker','time']])
x_test_smoker=ohe.transform(x_test[['smoker','time']])
x_test_smoker.shape
```

```python
si=SimpleImputer(strategy='mean')
x_train_size=si.fit_transform(x_train[['size']])
x_test_size=si.transform(x_test[['size']])
x_test_size.shape
```

```python
x_train_transformed=np.concatenate((x_train_totalbill,x_train_day,x_train_smoker,x_train_size),axis=1)
x_train_transformed.shape
```
