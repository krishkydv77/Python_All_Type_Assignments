```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('/content/Coffe_sales - Coffe_sales.csv')
df.head(5)
```

```python
df.isnull().sum()
```

```python
x=df.drop(columns=['money'])
y=df['money']
```

```python
x.head(5)
```

```python
from sklearn.preprocessing import LabelEncoder
lb=LabelEncoder()

x['Time_of_Day']=lb.fit_transform(x['Time_of_Day'])
x['Month_name']=lb.fit_transform(x['Month_name'])
x['Weekday']=lb.fit_transform(x['Weekday'])
x['cash_type']=lb.fit_transform(x['cash_type'])
x['coffee_name']=lb.fit_transform(x['coffee_name'])
x['Time_of_Day']=lb.fit_transform(x['Time_of_Day'])
x.head(5)
```

```python
x=x.drop(columns=['Date','Time'])
x.head(5)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=1)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train_sc=sc.fit_transform(x_train)
```

```python
x_train_sc_df=pd.DataFrame(x_train_sc,columns=x_train.columns)
x_train_sc_df.head(5)
```

```python
np.round(x_train_sc_df.describe(),2)
```

```python
df=pd.read_csv('/content/Coffe_sales - Coffe_sales.csv')
df.head(5)
```

```python
df=df.drop(columns=['Date','Time'])
df.head(5)
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe=OrdinalEncoder()
df['Time_of_Day']=oe.fit_transform(df[['Time_of_Day']])
df['Month_name']=oe.fit_transform(df[['Month_name']])
df['Weekday']=oe.fit_transform(df[['Weekday']])
df['cash_type']=oe.fit_transform(df[['cash_type']])
df['coffee_name']=oe.fit_transform(df[['coffee_name']])
df.head(5)
```

```python
x=df.drop(columns=['money'])
y=df['money']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
from sklearn.preprocessing import MinMaxScaler
ms=MinMaxScaler()
x_train_ms=ms.fit_transform(x_train)
```

```python
x_train_ms_df=pd.DataFrame(x_train_ms,columns=x_train.columns)
x_train_ms_df.head(5)
```

```python
np.round(x_train_ms_df.describe(),2)
```
