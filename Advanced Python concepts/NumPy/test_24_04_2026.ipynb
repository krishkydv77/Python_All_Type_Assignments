q5.
view= view in  change data,then change origanal data and refalect duplicate data

copy= copy in change data, then not change orignal data

```python
a=np.array([1,2,3,4,5,6,7])
a
```

```python
#view
a[3:6]=0
a
```

```python
#copy
a=np.array([1,2,3,4,5,6,7])
b=a[3:6].copy()
print("after change b data:",b)
b=0
print(a)
print("copy then change data:",b)
```
