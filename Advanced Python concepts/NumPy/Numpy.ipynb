hstack vs Vstack
- hstack --> It will add 2 or more array in horizontally.
- Vstack --> It will add 2 or more array in vertically.

```python
a=np.arange(1,10).reshape(3,3)
b=np.arange(11,20).reshape(3,3)
c=np.arange(21,30).reshape(3,3)
```

```python
np.hstack((a,b,c))
```

```python
np.vstack((a,b,c))
```
