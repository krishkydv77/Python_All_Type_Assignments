**Task 6: User Segmentation**

```python
print(df['daily_screen_time_minutes'].describe())
```

```python
#Addiction based segmentation
df['addiction_group'] = ''

df.loc[df['addiction_level_1_to_10'] <= 3, 'addiction_group'] = 'Low Addiction'
df.loc[(df['addiction_level_1_to_10'] > 3) & (df['addiction_level_1_to_10'] <= 7),
    'addiction_group'] = 'Moderate'

df.loc[df['addiction_level_1_to_10'] > 7, 'addiction_group'] = 'High Addiction'

#user type
df['user_type'] = ''
df.loc[df['daily_screen_time_minutes'] < 95, 'user_type'] = 'Low User'
df.loc[(df['daily_screen_time_minutes'] >= 95) & (df['daily_screen_time_minutes'] < 182),
    'user_type'] = 'Medium User'
df.loc[df['daily_screen_time_minutes'] >= 182, 'user_type'] = 'High User'

#Combined User Category
df['final_user_type:'] = df['user_type'] + " | " + df['addiction_group']

#count
print(df['final_user_type:'].value_counts())
```
