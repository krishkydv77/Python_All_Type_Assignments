```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

- **1. Univariate Analysis**

```python
df=pd.read_csv('/content/titanic - titanic.csv')
df
```

```python
df.columns
```

```python
df['Survived'].value_counts() # if we want to see acutal values then we use value_counts()
```

```python
sns.countplot(x=df['Survived'])
```

```python
df['Survived'].value_counts() # if we want to see acutal values then we use value_counts()
```

```python
df['Survived'].value_counts().plot(kind='bar')
```

```python
# if we want to find out percentage then use piechart
df['Survived'].value_counts().plot(kind='pie',autopct='%.2f')
```

- if we have numerical data then we use histogram because it finds the distribution.

```python
plt.hist(x=df['Age'])
plt.show()
```

- - Boxplot => for find out outliers

- 1. lower fence
- 2. 25% data
- 3. IQR(Inter Quartile Range(75%-25%))
- 4. 75% data
- 5. Upper fence

```python
sns.boxplot(x=df['Age'])
```
