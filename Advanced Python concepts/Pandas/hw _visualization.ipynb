```python
import matplotlib.pyplot as plt
import numpy as np

x = [1,2,3,4,5]
y = [5,7,6,8,7]

sizes = [50,100,150,200,250]

plt.figure(figsize=(6,4))

plt.scatter(x, y, c=y, s=sizes, cmap='viridis', edgecolor='black', alpha=0.7, label='Scatter Plot')

plt.title("Scatter Plot with Color Mapping")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.xlim(0,6)
plt.ylim(4,10)
plt.legend()

# labels
for i in range(len(x)):
    plt.text(x[i], y[i], y[i],ha='center')

# trend line
m, b = np.polyfit(x, y, 1)
plt.scatter(x, y)
plt.plot(x, m*np.array(x) + b, color='red', linestyle='--')   #y=mx+b

plt.text(1, max(y), f"y={m:.2f}x+{b:.2f}")
plt.colorbar()
plt.title("Advanced Scatter Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.xlim(0,6)
plt.ylim(4,10)
plt.legend()
plt.gca().set_facecolor('black')

plt.grid(True)


plt.tight_layout()
plt.show()
```

Q: What is autopct in pie chart?

"It shows the percentage value of each slice on the chart."

Q: Why use explode?

"To highlight a particular slice."

```python
# data
labels = ['Python','Java','C++','JS']
sizes = [40,25,20,15]

# explode option to highlight a slice (optional)
explode = (0.05, 0, 0, 0)  # Python slice thoda bahar
color=['red','green','yellow','blue']

# pie chart
plt.figure(figsize=(3,3))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=360, explode=explode,colors=color)

# title
plt.title("Programming Language Popularity")

plt.show()
```

tips dataset ka use karke bar plot banao jisme day x-axis aur total_bill ka average y-axis ho.

```python
sns.set_style('dark') #whitegrid/darkgrid/dark/white/tickes
sns.barplot(x='day',y='total_bill',data=tips,
            estimator=np.sum,# avg. kis base par nikalna ha sum/mean/mediam
            hue='sex', # kis category ke base par nikalna hai
            palette='dark', # color Popular Palette Options in Seaborn ()'pastel'/'deep' /'bright' /'dark'/'colorblind'/'coolwarm'
             # confidence interval
          )
sns.stripplot(x='day', y='total_bill', data=tips, color='black', alpha=0.6)  # actual data dekhne ke liye
plt.title('Tips Data')
plt.xlabel('Day')
plt.ylabel('Total Bill')

plt.show()
```

```python
sns.scatterplot(data=iris,x='sepal_length',y='petal_length',
                hue='species', #Color assign karne ke liye.
                style='species',    # shape by species
                s=100,
                alpha=0.5,
                palette='pastel')
sns.lmplot(x='sepal_length', y='petal_length', hue='species', data=iris)  #Add Regression Line (trendline)
plt.title('Iris Data')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.tight_layout()
plt.show()
```

diamonds dataset me correlation heatmap banao numeric features ke liye.

```python


# Dataset load
diamonds = sns.load_dataset('diamonds')

# Sirf numeric columns lo
numeric_data = diamonds.select_dtypes(include=['number'])

# Correlation matrix banao
corr_matrix = numeric_data.corr()
mask = np.triu(corr_matrix) # upper triangle hide karne ke liye bcoz dono triangle saome hote hai
# Heatmap
plt.figure(figsize=(10,8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f',mask=mask)

# Decoration
plt.title('Correlation Heatmap - Diamonds Dataset')
plt.show()
```

```python
corr = diamonds.select_dtypes(include='number').corr()

# Mask create
mask = np.triu(np.ones_like(corr, dtype=bool))

# Plot
plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap='coolwarm', mask=mask, linewidths=0.5)

plt.title('Correlation Heatmap (Lower Triangle)')
plt.show()
```

```python
data = {
    'month': ['Jan','Feb','Mar','Apr','May','Jun'],
    'sales': [200, 300, 250, 400, 350, 450],
    'profit': [20, 50, 30, 70, 60, 80],
    'region': ['North','South','East','West','North','South']
}

df = pd.DataFrame(data)
df
```

```python

fig1=px.line(df,x='month',y='sales',title='monthly sales trades',markers=True)
fig1.show()
```

```python
fig=px.bar(df,x='month',y='profit',color='region',barmode='group',title='monthly profit trades')
fig.show()
#
```

```python
fig3 = go.Figure()

fig3.add_trace(go.Scatter(
    x=df['month'],
    y=df['sales'],
    mode='lines+markers',
    name='Sales'
))

fig3.add_trace(go.Bar(
    x=df['month'],
    y=df['profit'],
    name='Profit'
))

fig3.update_layout(
    title='Sales vs Profit (Advanced)',
    xaxis_title='Month',
    yaxis_title='Amount'
)

fig3.show()
```

```python
from plotly.subplots import make_subplots

fig = make_subplots(rows=1, cols=2)

# px graph को go में convert करके add करना
fig.add_trace(
    go.Bar(x=df['month'], y=df['sales'], name='Sales'),
    row=1, col=1
)

fig.add_trace(
    go.Scatter(x=df['month'], y=df['profit'], mode='lines', name='Profit'),
    row=1, col=2
)

fig.update_layout(title="Mini Dashboard")

fig.show()
```
